````python
def produtoM3(lista):
    listaMenores = []
    i = 0
    res = 1
    while i < len(lista):
        if len(listaMenores) < 3:
            listaMenores.append(lista[i])
        else:
            x = 0
            while x < 3:
                if lista[i] < listaMenores[x] and not(lista[i] in listaMenores):
                    listaMenores[x] = lista[i] 
                x+=1
        i+=1
    print(listaMenores)
    for novoElem in listaMenores:
        res*=novoElem
    return res

print(produtoM3([12,3,7,10,12,8,9]))
# Resultado esperado: 168 = 3 * 7 * 8
````

#### c) Especifique uma função que dado um número inteiro positivo, repetidamente adiciona os seus dígitos até obter apenas um dígito que é retornado como resultado:

````python
# Input: 38
# Output: 2
# Explicação: 3 + 8 = 11, 1 + 1 = 2.

# Input: 777
# Output: 3
# Explicação: 7 + 7 + 7 = 21, 2 + 1 = 3.

def reduxInt(n):
    resfinal = ''
    while len(resfinal) != 1:
        if resfinal == '':
            res = 0
            numstr = str(n)
            for num in numstr:
                res+=int(num)
        else:
            res = 0
            for num in resfinal:
                res+=int(num)
        resfinal = str(res)
    return resfinal

print(reduxInt(38))
print(reduxInt(777))
print(reduxInt(1000))
print(reduxInt(444))
````

### tpc-3. A Rede Social

Considere que a informação sobre uma rede social está armazenada numa lista de dicionários.

Cada dicionário, correspondente a um _post_ e tem chaves `id`, `conteudo`, `autor`, `dataCriacao` e `comentarios`.
Por sua vez, `comentarios` é uma lista de dicionários com chaves `comentario` e `autor`.

Considere o seguinte exemplo:

``` 
    MyFaceBook = [{

    tem de realizar...',

        ]},
```

````python
myFaceBook = [
    {   'id': 'p1', 
        'conteudo': 'A tarefa de avaliação é talvez a mais ingrata das tarefas que um professor tem de realizar...', 
        'autor': 'jcr', 
        'dataCriacao': '2023-07-20', 
        'comentarios': [
            {
                'comentario': 'Completamente de acordo...',
                'autor': 'prh'
            },
            {
                'comentario': 'Mas há quem goste...',
                'autor': 'jj'
            }
    ]},
    {   'id': 'p2',
        'conteudo': 'Isto é um post de teste',
        'autor': 'af',
        'dataCriacao':'2026-11-02',
        'comentarios': [
            {
                'comentario': 'Isto é um comentário de teste',
                'autor': 'qdc'
            },
            {
                'comentario': 'Isto é outro teste',
                'autor': 'jvr'
            }
    ]},
    {   'id':'p3',
        'conteudo': 'Isto é um outro post do mesmo autor',
        'autor': 'rf',
        'dataCriacao':'2026-11-02',
        'comentarios': []
    }]
````

#### a) `quantosPost`, que indica quantos posts estão registados:

````python
def quantosPost(redeSocial):
    contador = 0
    for posts in redeSocial:
        for chave in posts.items():
           if chave == 'id':
               contador+=1
    return contador

print(quantosPost(myFaceBook))
````

#### b)  `postsAutor`, que devolve a lista de posts de um determinado autor:

````python
def postsAutor(redeSocial, autor):
    res = [] 
    for posts in redeSocial:
        if posts['autor'] == autor:
            res.append(posts['conteudo'])
    return res

print(postsAutor(myFaceBook,'jcr'))
print(postsAutor(myFaceBook,'rf'))
````

#### c) `autores`, que devolve a lista de autores de posts ordenada alfabeticamente:

````python
def ordenaLista (a):
    return a[0][0]

def autores(redeSocial):
    res = []
    for posts in redeSocial:
        if not(posts['autor'] in res):
            res.append(posts['autor'])
    res.sort(key=ordenaLista)
    return res

print(autores(myFaceBook))
````

#### d) `insPost`, que acrescenta um novo post à rede social a partir dos parâmetros recebidos e devolve a nova rede social. 
    
O campo `id` deverá ser calculado a partir dos já existentes, por exemplo, se a rede tiver posts com id `p1`, `p2` e `p3`, o novo `id` deverá ser `p4`.

````python
def insPost(redeSocial, conteudo, autor, dataCriacao, comentarios):
    novoId = f'p{int(redeSocial[-1]['id'][-1])+1}'
    post = {
        'id': novoId,
        'conteudo': conteudo,
        'autor': autor,
        'dataCiacao': dataCriacao,
        'comentarios': comentarios
    }
    redeSocial.append(post)
    return redeSocial

insPost(myFaceBook,'Adição de um post de teste','zbf','2026-11-02',[{'comentario': 'Teste adição comentário', 'autor': 'jar'}])
insPost(myFaceBook,'Adição de um post de teste2','bzg','2026-11-02',[{'comentario': 'Outro teste adição comentário', 'autor': 'bar'}])
````

#### e)  `remPost`, que remove um post da rede, correspondente ao `id` recebido.

````python
def remPost(redeSocial, id):
    contador = 0
    for posts in redeSocial:
        if id != posts['id']:
            contador+=1
        else:
            redeSocial.pop(contador)
    return redeSocial

remPost(myFaceBook,'p4')
print(myFaceBook)
````

#### f) `postsPorAutor`, que devolve uma distribuição de posts por autor (à semelhança do que foi feito nas aulas).

````python
def postsPorAutor(redeSocial):
    res = {}
    for posts in redeSocial:
        if not(posts['autor'] in res):
            res[posts['autor']] = [posts['conteudo']]
        else:
            res[posts['autor']].append(posts['conteudo'])
    return res

postsPorAutor(myFaceBook)
````

#### g) `comentadoPor`, que recebe um autor e devolve a lista de posts comentados por esse autor.

````python
def comentadoPor(redeSocial, autor):
    res = []
    for posts in redeSocial:
        for comentarios in posts['comentarios']:
            if comentarios['autor'] == autor:
                res.append(comentarios['comentario'])
    return res

print(comentadoPor(myFaceBook,'jj'))
````
